```python
import pandas as pd 
import requests
import io
#url = "https://populationtoday.com/"
df=pd.read_csv("population_data.csv",encoding="utf-8")
print("Total missing values",df.isnull().sum())
print("Total Dublicate values",df.duplicated().sum())
print(f"the total population in the world in 2026 is: {df['Population 2026'].sum()}")
min_population=df['Population 2026'].min()
Max_population=df['Population 2026'].max()
Max_country=df.loc[df['Population 2026']==Max_population,'Country (or dependency)']
print("the Country with the largest population is",Max_country.to_string(index=False))
Min_country=df.loc[df['Population 2026']==min_population,'Country (or dependency)']
print("the Country with the Smallets population is",Min_country.to_string(index=False))
min_growth_rate=df['Yearly Change'].min()
max_growth_rate=df['Yearly Change'].max()
Max_country=df.loc[df['Yearly Change']==max_growth_rate,'Country (or dependency)']
print("Country with Highest Growth Rate  is",Max_country.to_string(index=False),max_growth_rate)
Min_country=df.loc[df['Yearly Change']==min_growth_rate,'Country (or dependency)']
print("Country with lowest Growth Rate is",Min_country.to_string(index=False),min_growth_rate)
print(df.info())
print(df.describe())
print(df.head)
```
